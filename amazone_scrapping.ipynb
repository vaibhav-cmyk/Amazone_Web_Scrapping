```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import pandas as pd
import time

# Setup Selenium
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get("https://www.flipkart.com/clothing-and-accessories/pr?sid=clo&p%5B%5D=facets.ideal_for%255B%255D%3DMen")

time.sleep(5)  # wait for JS to load

name_list = []
price_list = []
discount_list = []

while True:
    soup = BeautifulSoup(driver.page_source, "lxml")

    names = soup.find_all("a", class_="WKTcLC BwBZTg")
    prices = soup.find_all("div", class_="Nx9bqj")
    discounts = soup.find_all("div", class_="UkUFwK")

    for n in names:
        name_list.append(n.text)
    for p in prices:
        price_list.append(p.text)
    for d in discounts:
        discount_list.append(d.text)

    # Pagination
    next_button = driver.find_elements("class name", "_9QVEpD")
    if next_button:
        next_button[0].click()
        time.sleep(5)
    else:
        break

driver.quit()

# Save to CSV
df = pd.DataFrame({
    "Name": name_list,
    "Price": price_list,
    "Discount": discount_list
})
df.to_csv("flipkart_scrap.csv", index=False)
```
